```python
import pandas as pd

matches = pd.read_csv("../Dataset/matches.csv")
deliveries = pd.read_csv("../Dataset/deliveries.csv")

print(matches.shape)
print(deliveries.shape)

matches.head()
```

```python
print(matches.columns)

print("\n" + "="*50 + "\n")

print(deliveries.columns)
```

```python
total_matches = matches['id'].nunique()

total_teams = len(set(matches['team1']).union(set(matches['team2'])))

total_runs = deliveries['batsman_runs'].sum()

total_sixes = deliveries[deliveries['batsman_runs'] == 6].shape[0]

total_fours = deliveries[deliveries['batsman_runs'] == 4].shape[0]

print("Total Matches:", total_matches)
print("Total Teams:", total_teams)
print("Total Runs:", total_runs)
print("Total Sixes:", total_sixes)
print("Total Fours:", total_fours)
```

```python
team_wins = matches['winner'].value_counts()

print(team_wins.head(10))
```

```python
top_batsmen = (
    deliveries.groupby('batsman')['batsman_runs']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

top_batsmen.head(10)
```

```python
top_bowlers = (
    deliveries[deliveries['player_dismissed'].notna()]
    .groupby('bowler')['player_dismissed']
    .count()
    .sort_values(ascending=False)
    .reset_index()
)

top_bowlers.head(10)
```

```python
most_sixes = (
    deliveries[deliveries['batsman_runs'] == 6]
    .groupby('batsman')
    .size()
    .sort_values(ascending=False)
    .reset_index(name='Sixes')
)

most_sixes.head(10)
```

```python
most_fours = (
    deliveries[deliveries['batsman_runs'] == 4]
    .groupby('batsman')
    .size()
    .sort_values(ascending=False)
    .reset_index(name='Fours')
)

most_fours.head(10)
```

```python
team_wins.reset_index().to_csv(
    '../dataset/team_wins.csv',
    index=False
)

top_batsmen.to_csv(
    '../dataset/top_batsmen.csv',
    index=False
)

top_bowlers.to_csv(
    '../dataset/top_bowlers.csv',
    index=False
)

most_sixes.to_csv(
    '../dataset/most_sixes.csv',
    index=False
)

most_fours.to_csv(
    '../dataset/most_fours.csv',
    index=False
)
```
